**Data Cleaning & Standardization**

```python
def clean_state_text(x):
    if pd.isna(x):
        return x
    x = x.strip()
    x = re.sub(r'\s+', ' ', x)
    x = re.sub(r'\s*&\s*', ' and ', x)
    x = x.title()
    return x

def standardize_state(x):
    return STATE_MAPPING.get(x, x)


def clean_district_text(x):
    if pd.isna(x):
        return x

    x = str(x)
    x = re.sub(r'[＊*?•]', '', x)
    x = re.sub(r'[-–—]', ' ', x)
    x = re.sub(r'\s*&\s*', ' and ', x)
    x = re.sub(r'[()]', '', x)
    x = re.sub(r'\s+', ' ', x)

    return x.strip().title()

def standardize_district(x):

    return DISTRICT_MAP.get(x, x)
```

```python
for df in [df_bio, df_demo, df_enroll]:
    df['state'] = df['state'].apply(clean_state_text)
    df['district'] = df['district'].apply(clean_district_text)
```

Raw pincode-level data is too granular.

For insights, we usually aggregate to:

District–Month

or State–Month

Example: District–Month aggregation

```python
bio_agg = df_bio.groupby(['date', 'state', 'district'])[
    ['bio_age_5_17', 'bio_age_17_']
].sum().reset_index()

demo_agg = df_demo.groupby(['date', 'state', 'district'])[
    ['demo_age_5_17', 'demo_age_17_']
].sum().reset_index()

enroll_agg = df_enroll.groupby(['date', 'state', 'district'])[
    ['age_0_5', 'age_5_17', 'age_18_greater']
].sum().reset_index()
```

```python
bio_agg
```

For each (date, state, district):

Keep the row only if it exists in BOTH:

enrollment data

biometric data

❌ If a district has enrollment data but no biometric data that day, it is dropped.

```python
df_final = enroll_agg.merge(
    bio_agg,
    on=['date', 'state', 'district'],
    how='inner'
).merge(
    demo_agg,
    on=['date', 'state', 'district'],
    how='inner'
)
```
